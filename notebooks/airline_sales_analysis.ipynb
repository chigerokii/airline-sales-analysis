#### <a class="anchor" id="6.2"></a><div class="alert alert-block alert-success">Задание 3.2

Исследуйте выручку, примите решение об исключении выбросов для дальнейшего исследования, руководствуясь минимизацией потерь в данных.

<b><a class="anchor" id="6.2.1"></a><div class="alert alert-block alert-success">Исследование выручки</b>

```python
#рассчитаем с помощью describe статистические показатели
data['revenue_amount'].describe()
```

```python
#построим boxplot для анализа выбросов
sns.boxplot(data['revenue_amount']).set(title='Ящик с усами для выручки', ylabel='Выручка')
plt.show()
```

```python
#построим график плотности, чтобы проверить нормальное ли распределение
sns.kdeplot(data['revenue_amount']).set(title='Плотность распределения выручки', xlabel='Выручка', ylabel='Плотность')
plt.show()
```

```python
#по той логике построим Q-Q plot 
plt.figure(figsize=(10,6))
st.probplot(data['revenue_amount'], dist='norm', plot=plt)
plt.show()
```

```python
#проведем статистический тест Шапиро-Уилка
stat, p = st.shapiro(data['revenue_amount'])
alpha = 0.05
if p < alpha:
    print(p, 'Отвергаем нулевую гипотезу. Генеральная совокупность НЕ имеет нормального распределения')
else:
    print(p, 'Не получилось отвергнуть нулевую гипотезу. Генеральная совокупность имеет нормальное распределение')
```

```python
#создаем функцию, которая поможет нам с выбросами
def ejection (col):
    q75 = np.percentile(data[col], 75) # считаем все элементы диаграммы размаха
    q25 = np.percentile(data[col], 25)
    iqr = q75 - q25
    ejection_h = q75 + 1.5*iqr
    ejection_l = q25 - 1.5*iqr
    ejection_count = data[col][(data[col] > ejection_h) | (data[col] < ejection_l)].count() # считаем число выбросов
    ejection_perc = ((ejection_count / data.shape[0]) * 100).round(2) # считаем долю числа выборосов в общем датасете
    return(ejection_h, ejection_l, ejection_count, ejection_perc)
```

```python
ejection('revenue_amount') #результат говорит нам о наличии экстремальных выбросов
```

```python
#определяем границу значений при перцинтиле = 99.9
revenue_amount_ejection_h = np.percentile(data['revenue_amount'], 99.9)
revenue_amount_ejection_h
```

```python
#считаем количество экстремальных выбросов
revenue_amount_count = data['revenue_amount'][(data['revenue_amount'] > revenue_amount_ejection_h)].count()
revenue_amount_count 
```

```python
#считаем процент от общего датасета экстремальных выбросов
ejection_perc = ((revenue_amount_count / data.shape[0]) * 100).round(2)
ejection_perc
```

```python
#строим гистограмму для наглядности
data['revenue_amount'].hist()
```

```python
#перезаписываем датасет без экстремальных выбросов
data = data[(data['revenue_amount'] < revenue_amount_ejection_h)].reset_index(drop=True)
```

```python
#создаем датафрейм с выручкой зимой
winter_revenue_amount = data[data['season'] == 'Зима'][['season', 'revenue_amount']].reset_index(drop=True)
winter_revenue_amount.head(1)
```

Определите топ-10 популярных направлений перелета (город-город) отдельно для ВМЛ и МВЛ. Изучите выручку для этих направлений. Для этого добавьте в датасет сведения о городах, используя данные из датасета: https://raw.githubusercontent.com/jpatokal/openflights/master/data/airports.dat

```python
#создаем список с названиями столбцов для датасета из гитхаба
column_names = ['Airport Name', 'Location', 'Country', 'IATA_Code', 'ICAO_Code', 'Latitude', 'Longitude', 'Altitude', 'Timezone', 'DST', 'Timezone Region', 'Type', 'Source']
```

```python
#считываем ссылку и создаем датафрейм 
df = pd.read_csv('https://raw.githubusercontent.com/jpatokal/openflights/master/data/airports.dat', header=None, names=column_names)
df.head(1)
```

```python
#cоздаем словарь, где ключ - код аэропорта, а значение - название города
dct_code = {'PAR': 'Paris', 'KUF': 'Samara', 'BAK': 'Baku', 'ROM': 'Roma', 'VVO': 'Vladivostok', 'OVB': 'Novosibirsk', 
           'SEL': 'Seoul', 'MOW': 'Moscow', 'MRV': 'Mineralnye Vody', 'BER': 'Berlin', 'GRV': 'Groznyj', 'MCX': 'Makhachkala',
           'NQZ': 'Nursultan', 'KRR': 'Krasnodar', 'VOG': 'Volgograd', 'OHH': 'Oha', 'ASF': 'Astrakhan', 'TOX': 'Tobolsk',
           'MIL': 'Milan', 'TYO': 'Tivat', 'LON': 'London', 'IAR': 'Yaroslavl', 'KHV': 'Khabarovsk', 'BJS': 'Beijing', 
           'BKK': 'Bangkok', 'AER': 'Sochi', 'DUS': 'Düsseldorf'}
```

```python
#создаем функцию, которая будет возращать название города
def city_code(code):
    result = df[df['IATA_Code'] == code]['Location']
    if not result.empty:
        return result.iloc[0]
    elif code in dct_code.keys():
        return dct_code[code]
```

```python
#подсчитываем для МВЛ сумму выручки и количество покупок для всех направлений
mvl_revenue_amount = data[data['route_flight_type'] == 'МВЛ'].groupby(['direction'])['revenue_amount'].agg(['sum', 'count']).sort_values(by='count', ascending=False).reset_index()
mvl_revenue_amount.head(1)
```

```python
#визуализируем 
plt.figure(figsize=(10, 6))
sns.barplot(vvl_revenue_amount.sort_values(by='sum', ascending=False).head(10), y='direction', x='sum', orient='h').set(title='Распределение суммы выручки среди топ-10 популярных направлений для ВВЛ', xlabel='Выручка', ylabel='Направление')
plt.show()
```

#### <a class="anchor" id="6.6.1"></a><div class="alert alert-block alert-success">Исследование корреляции выручки с другими признаками

```python
#переводим все данные в числовой числовой код, так как имеются категориальные переменные
data_factorize = data.apply(lambda x: pd.factorize(x)[0])
data_factorize.head()
```

```python
#строим матрицу корреляций, используя ранговую корреляцию Кендалла, который позволяет работать с категориальными переменными, их проранжировав
corr_matrix = data_factorize.corr(method='kendall')
corr_matrix.head(1)
```

```python
#визуализируем
plt.figure(figsize=(15, 5))
sns.heatmap(corr_matrix, annot=True)
plt.title('Матрица корреляций')
plt.show()
```
